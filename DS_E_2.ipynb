# Q1) a. Breadth-First Search (BFS)


```python
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    bfs_result = []

    while queue:
        node = queue.popleft()

        if node not in visited:
            bfs_result.append(node)
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)
    return bfs_result

# Perform BFS starting from node 'A'
bfs_result = bfs(graph, 'A')
print(f"BFS Traversal: {bfs_result}")
```

# Q1) b.  Depth-First Search (DFS)


```python
def dfs(graph, start):
    visited = set()
    stack = [start]
    dfs_result = []

    while stack:
        node = stack.pop()

        if node not in visited:
            dfs_result.append(node)
            visited.add(node)
            # Push neighbors onto the stack in reverse order to explore in correct order
            # (e.g., if A: [B, C], we want to visit B before C, so push C then B)
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    return dfs_result

# Perform DFS starting from node 'A'
dfs_result = dfs(graph, 'A')
print(f"DFS Traversal: {dfs_result}")
```

# Q2) A* Algorithm




```python
import heapq

def a_star(graph, start, goal, h):
    # g_score: cost from start to current node
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    # f_score: g_score + heuristic (estimated total cost from start to goal through current node)
    f_score = {node: float('inf') for node in graph}
    f_score[start] = h[start]

    # priority queue of (f_score, node)
    open_set = [(f_score[start], start)]

    # came_from: tracks the path reconstruction
    came_from = {}

    while open_set:
        current_f_score, current_node = heapq.heappop(open_set)

        if current_node == goal:
            # Reconstruct path
            path = []
            while current_node in came_from:
                path.append(current_node)
                current_node = came_from[current_node]
            path.append(start)
            return path[::-1], g_score[goal] # Return reversed path and total cost

        for neighbor, weight in graph[current_node].items():
            tentative_g_score = g_score[current_node] + weight

            if tentative_g_score < g_score[neighbor]:
                came_from[neighbor] = current_node
                g_score[neighbor] = tentative_g_score
                f_score[neighbor] = g_score[neighbor] + h[neighbor]
                # Add to open_set if not already there, or if we found a better path
                # The heapq handles updating priorities if an item is re-added with a lower f_score
                heapq.heappush(open_set, (f_score[neighbor], neighbor))

    return None, float('inf') # No path found
```

#### Example Usage

Let's apply the A* algorithm to a sample weighted graph with a defined heuristic function to find the shortest path from a start node to a goal node.

```python
# Define a sample graph with weights
graph = {
    'A': {'B': 6, 'D': 4},
    'B': {'A': 6, 'E': 7, 'C': 5},
    'C': {'B': 5, 'E': 8, 'F': 5},
    'D': {'A': 4, 'E': 6},
    'E': {'B': 7, 'C': 8, 'D': 6, 'F': 4},
    'F': {'C': 5, 'E': 4}
}

# Define a heuristic function (estimated cost from each node to the goal 'F')
h = {
    'A': 10,
    'B': 8,
    'C': 3,
    'D': 9,
    'E': 4,
    'F': 0
}

start_node = 'A'
goal_node = 'F'

path, cost = a_star(graph, start_node, goal_node, h)

if path:
    print(f"Shortest path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}")
```

# Q3) Minimax Algorithm

```python
import math

def minimax(node, depth, maximizingPlayer):
    # Base case: if depth is 0 or node is a terminal node
    # In this example, terminal nodes are those without children (leaf nodes)
    if depth == 0 or not node.children:
        return node.value

    if maximizingPlayer:
        max_eval = -math.inf
        for child in node.children:
            eval = minimax(child, depth - 1, False)
            max_eval = max(max_eval, eval)
        return max_eval
    else:
        min_eval = math.inf
        for child in node.children:
            eval = minimax(child, depth - 1, True)
            min_eval = min(min_eval, eval)
        return min_eval

# Helper class to represent a node in the game tree
class Node:
    def __init__(self, value=None, children=None):
        self.value = value  # Only for leaf nodes
        self.children = children if children is not None else []

    def add_child(self, child_node):
        self.children.append(child_node)
```

#### Example Usage: Simple Game Tree

Let's construct a simple game tree to demonstrate the Minimax algorithm. We'll define a tree with a depth of 2, where the maximizing player goes first.

```python
# Construct a sample game tree
# Leaf nodes with values
leaf1 = Node(value=3)
leaf2 = Node(value=5)
leaf3 = Node(value=2)
leaf4 = Node(value=9)
leaf5 = Node(value=12)
leaf6 = Node(value=5)
leaf7 = Node(value=23)
leaf8 = Node(value=2)

# Nodes at depth 1
nodeB = Node(children=[leaf1, leaf2])
nodeC = Node(children=[leaf3, leaf4])
nodeD = Node(children=[leaf5, leaf6])
nodeE = Node(children=[leaf7, leaf8])

# Root node (depth 0)
root = Node(children=[nodeB, nodeC, nodeD, nodeE])

# Assume the root is the current state, depth to search is 2
# True indicates the current player (at root) is the maximizing player
optimal_value = minimax(root, 2, True)

print(f"The optimal value for the maximizing player is: {optimal_value}")

# To find the actual optimal move, we'd typically iterate through the root's children
# and apply minimax to each to see which child leads to the optimal value.
# Let's extend the example to show the best first move.

best_move_value = -math.inf
best_move = None

print("\nDetermining the best first move:")
for i, child_node in enumerate(root.children):
    # Simulate opponent's turn (minimizing player)
    current_move_value = minimax(child_node, 1, False) # depth 1, opponent is minimizing
    print(f"  Move to child {chr(ord('B') + i)} leads to value: {current_move_value}")
    if current_move_value > best_move_value:
        best_move_value = current_move_value
        best_move = chr(ord('B') + i) # Assuming 'B', 'C', 'D', 'E' as child names

print(f"The best first move for the maximizing player is to node {best_move}, leading to a final score of {best_move_value}.")
```
